## Evaluating Derivatives and Gradients

```python
# calculating slope/derivative of function at point x. 
# We move in direction opposite to derivative(tagent to curve) to find least value of x

def Derivative(x):
    d = 2*(x-2)
    return d

#General method to calculate derivative...very off result
def dfdx(x):
    h = 10^-4
    dfdx = (f(x+h) - f(x))/h
    return dfdx
```

```python
# calculating gradient of function at point x.
# We move in direction opposite to gradient (normal to curve) to find least value of x
import numpy as np
def Gradient(x, y):
    # gradient is the matrix of derivatives of the function
    # so, define a 1D matrix... using a 1D array

    der_x = 10*x - 30 + 4*y   # derivative wrt x
    der_y = 2*y - 8 + 4*x    # derivative wrt y
    grad_vector = np.array([der_x, der_y])

    return grad_vector


def Grad_Norm(x, y):
    der_x = 10*x - 30 + 4*y   # derivative wrt x
    der_y = 2*y - 8 + 4*x    # derivative wrt y
    grad_norm = (der_x**2 + der_y**2)**(1/2)

    return grad_norm
```

```python
# calculating gradient of function at point x.
# We move in direction opposite to gradient (normal to curve) to find least value of x

def Gradient_(x, y):
    
    der_x = 24*x**2 - 6  # derivative wrt x
    der_y = 10*y   # derivative wrt y
    grad_vector = np.array([der_x, der_y])

    return grad_vector

def Grad_Norm_(x, y):
    der_x = 24*x**2 - 6  # derivative wrt x
    der_y = 10*y   # derivative wrt y
    grad_norm = (der_x**2 + der_y**2)**(1/2)

    return grad_norm
```

## Implementing Gradient Descent

```python
import numpy as np
import matplotlib.pyplot as plt

def GradientDescent(n):
    x = []
    y = []
    f1 = 0.0
    g1 = 0.0
    h1 = 0.0
    i = 0
    
    # define learning rate, take input from user = e
    e = float(input("Enter learning rate"))
    print("Learning Rate is: ", e)

    # initialise x and y to start off iterations
    x1 = 0.0
    y1 = 0.0
    v = np.array([x1, y1])   # vector array for x1 and y1

# else if statements have been used cause for different functions there are different functions for their gradient or derivative 
    if(n==2):
        grad_vector = Gradient(x1, y1) # gives gradient of v
        grad_norm = Grad_Norm(x1, y1)   

        while (grad_norm > 0.0001):
            v = v - e * grad_vector      # updates v or [x1, y1]
            grad_vector = Gradient(v[0], v[1])
            grad_norm = Grad_Norm(v[0], v[1])
            g1 = g(v[0],v[1])
            y.append(g1)
            i = i+1
            x.append(i)

        x = np.array(x)
        y = np.array(y)

        #plot of objective function v/s iterations
        plt.scatter(x, y)
        plt.xlabel("no. of iterations")
        plt.ylabel("g(x)")
        plt.title("Plot of g(x) v/s no. of iterations")
        plt.grid()
        plt.show()
            

        return v
    
    elif(n==1):
        while (abs(Derivative(v[0])) > 0.001):
            v[0] = v[0] - e*(Derivative(v[0]))
            f1 = f(v[0])
            y.append(f1)
            i = i+1
            x.append(i)

        x = np.array(x)
        y = np.array(y)

        #plot of objective function v/s iterations
        plt.scatter(x, y)
        plt.xlabel("no. of iterations")
        plt.ylabel("f(x) = $(x-2)^2$")
        plt.title("Plot of f(x) v/s no. of iterations")
        plt.grid()
        plt.show()

        return v[0]
    
    elif(n==3):
        grad_vector = Gradient_(x1, y1) # gives gradient of v
        grad_norm = Grad_Norm_(x1, y1)   

        while (grad_norm > 0.0001):
            v = v - e * grad_vector      # updates v or [x1, y1]
            grad_vector = Gradient_(v[0], v[1])
            grad_norm = Grad_Norm_(v[0], v[1])
            h1 = h(v[0],v[1])
            y.append(h1)
            i = i+1
            x.append(i)

        x = np.array(x)
        y = np.array(y)

        #plot of objective function v/s iterations
        plt.scatter(x, y)
        plt.xlabel("no. of iterations")
        plt.ylabel("h(x)")
        plt.title("Plot of h(x) v/s no. of iterations")
        plt.grid()
        plt.show()
        return v

    # end function, return final value of x, y i.e. the optimal value
```
